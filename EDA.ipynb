```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# 1. DATA LOADING & INSPECTION
# ---------------------------------------------------------
# Membuat dummy dataset / memuat dataset kantin
data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None, 'Nasi Goreng'],
    'harga': [12000, 4000, 10000, 4000, 8000, 12000],
    'terjual': [23, 40, None, 35, 18, 23]
}
df = pd.DataFrame(data_kantin)

print("=== 5 BARIS PERTAMA (HEAD) ===")
print(df.head())
print("\n=== INFORMASI DATASET (INFO) ===")
print(df.info())
print("\n=== STATISTIK RINGKAS (DESCRIBE) ===")
print(df.describe())
print("\n=== DIMENSI DATASET (SHAPE) ===")
print(df.shape)

# ---------------------------------------------------------
# 2. DATA CLEANING
# ---------------------------------------------------------
# A. Menangani Missing Value
df['terjual'] = df['terjual'].fillna(0)  # Isi penjualan kosong dengan 0
df = df.dropna(subset=['menu'])          # Hapus baris jika menu tidak diketahui

# B. Menangani Duplikat
df = df.drop_duplicates()

# C. Memastikan Tipe Data Sesuai
df['harga'] = df['harga'].astype(int)
df['terjual'] = df['terjual'].astype(int)

print("\n=== INFORMASI SETELAH CLEANING ===")
print(df.info())

# ---------------------------------------------------------
# 3. DATA MANIPULATION
# ---------------------------------------------------------
# A. Kolom Turunan
df['total_pendapatan'] = df['harga'] * df['terjual']

# B. Filtering (Menu dengan penjualan > 20)
menu_laris = df[df['terjual'] > 20]

# C. Sorting (Diurutkan berdasarkan pendapatan tertinggi)
df_sorted = df.sort_values(by='total_pendapatan', ascending=False)

# D. Groupby / Agregasi
ringkasan_menu = df.groupby('menu')[['terjual', 'total_pendapatan']].sum().reset_index()

print("\n=== DATASET SETELAH MANIPULASI (SORTED) ===")
print(df_sorted)
print("\n=== RINGKASAN AGREGASI PER MENU ===")
print(ringkasan_menu)

# ---------------------------------------------------------
# 4. SIMPAN DATASET BERSIH
# ---------------------------------------------------------
df_sorted.to_csv('dataset_bersih.csv', index=False)
print("\nDataset bersih berhasil disimpan sebagai 'dataset_bersih.csv'")
```
