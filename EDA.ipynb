```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

```python
import pandas as pd
data_kantin = {
 'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
 'harga': [12000, 4000, 10000, 4000, 8000],
 'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Analisis 2:

- Tabel yang terlihat none/kosong adalah menu di indeks ke 4 dan terjual di indeks ke 2.

-  Jika data ini tidak dibersihkan terlebih dahulu maka akan menyebabkan perhitungan tidak akurat, dan bisa saja terjadi error.

```python
df = pd.read_csv('data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Analisis 3:

- Kolom yang jumlah non-null nya lebih sedikit adalah kolom terjual dan kolom menu.

- Artinya terdapat data kosong (*missing values*) pada kolom tersebut, sehingga perlu dibersihkan sebelum dianalisis.


```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Analisis 4:

- Kenapa kolom terjual diisi? Karena ada menunya tapi tidak ada yang beli, jadi diisi 0.

- Kenapa menu kosong dihapus?
Karena ada menunya tapi ga ada yang beli. Daripada data tidak jelas, jadi lebih baik dihapus saja.


```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Analisis 5:

- Jumlah baris setelah  *drop_duplicates()* berkurang sebanyak baris yang terduplikasi.

- Mengapa penting memastikan tipe data sudah benar? Untuk mencegah error dan menghemat memori.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Analisis 6:

- Menu yang menghasilkan *total_pendapatan* tertinggi adalah Es Teh dan Nasi Goreng.

- Keputusan Bisnis: Menjaga pasokan bahan baku menu terlaris agar tidak kehabisan, serta mengevaluasi menu yang kurang laku.
